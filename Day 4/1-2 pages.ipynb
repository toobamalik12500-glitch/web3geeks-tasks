Week 1 Day 4 – Model Tuning, Diagnostics and Final Performance Report
Introduction

In this task, different machine learning models were trained and tuned to find the best performing model. The main goal was to improve the model performance and check whether the model gives reliable predictions.

The dataset contains 48,842 rows and 15 columns. The data was divided into training and testing sets. The training data contains 39,073 rows, while the test data contains 9,769 rows. A random state of 42 was used to make the results reproducible.

Model Tuning

Three models were tested:

Logistic Regression
Random Forest
Gradient Boosting

GridSearchCV with 5-fold Stratified Cross-Validation was used for hyperparameter tuning.

The best results were:

Model	Best CV Score
Logistic Regression	0.8569
Random Forest	0.8614
Gradient Boosting	0.8729

Gradient Boosting achieved the highest cross-validation score of 0.8729, so it was selected as the final model.

The best parameters for Gradient Boosting were:

Learning rate = 0.1
Max depth = 5
Number of estimators = 200
the model's predicted probabilities were reasonably good.

Different classification thresholds were also tested:

Threshold	Precision	Recall	F1 Score
0.3	0.6615	0.8139	0.7298
0.5	0.7923	0.6591	0.7196
0.7	0.8975	0.4795	0.6250

The threshold 0.3 gave the highest F1 Score of 0.7298. Therefore, 0.3 was selected when the goal is to maximize the F1 Score.

Final Test Performance

The final tuned Gradient Boosting model was evaluated on the untouched test set.

Metric	Score
Accuracy	0.8771
Precision	0.7923
Recall	0.6591
F1 Score	0.7196
Brier Score	0.0870

The model achieved an accuracy of about 87.7% on the test set. The precision was about 79.2%, while the recall was about 65.9%. The F1 Score was 71.96%.
 Final Model and Production Behavior

Gradient Boosting was selected as the final model because it achieved the highest cross-validation score among the three tested models.

The final preprocessing and model pipeline was saved as:

final_model.pkl

Saving the complete pipeline makes it easier to use the model later because the preprocessing steps and model are saved together.

In production, the model is expected to perform well on data that is similar to the training data. However, performance can change when new real-world data is different from the training data. Therefore, the model should be monitored and evaluated regularly.

Conclusion

Overall, Gradient Boosting performed better than Logistic Regression and Random Forest. After hyperparameter tuning, it achieved the best cross-validation score of 0.8729. On the final test set, it achieved 87.7% accuracy and an F1 Score of 0.7196. The Brier Score of 0.0870 also showed reasonably good probability predictions.

The model was successfully saved as final_model.pkl, completing the final model development and evaluation process.
