 Day 3 Summary: Feature Engineering, Cross-Validation & Model Comparison

 Engineered Features

In this task, six new features were created to add useful information to the Adult Census Income dataset.

| Feature                                             | Type        | Purpose                                                      |
| --------------------------------------------------- | ----------- | ------------------------------------------------------------ |
| age_group                                         | Categorical | Groups people into different age ranges.                     |
|`hours_group                                       | Categorical | Groups working hours into Low, Normal, High and Very High.   |
| has_capital_gain`mark down preview use ro k editor | Binary      | Shows whether a person has any capital gain.                 |
| log_capital_gain                                  | Numeric     | Reduces the effect of very large capital gain values.        |
| higher_education                                 | Binary      | Shows whether a person has a higher level of education.      |
| education_hours_interaction                      | Numeric     | Combines education level and working hours into one feature. |

These features were added to the preprocessing pipeline so they are created automatically when the model is trained.

 Cross-Validation and Model Comparison

Stratified 5-Fold Cross-Validation was used to compare three machine learning models:

 Logistic Regression
 Random Forest
 Gradient Boosting

The results were:

| Model               |    Mean F1 | F1 Std | Mean ROC AUC | ROC AUC Std |
| ------------------- | ---------: | -----: | -----------: | ----------: |
| Logistic Regression |     0.6732 | 0.0042 |       0.9129 |      0.0022 |
| Random Forest       |     0.6701 | 0.0063 |       0.9010 |      0.0021 |
| Gradient Boosting   | **0.6844** | 0.0074 |   **0.9203** |      0.0018 |

Gradient Boosting gave the highest F1 score and ROC AUC. Therefore, it was the best-performing model among the three models tested.

The boxplots were also used to compare the model scores across the five folds.

. Statistical Test

A paired t-test was performed between Gradient Boosting and Logistic Regression because they were the top two models based on F1 score.

 T-statistic: **5.763**
 P-value: **0.0045**

The p-value is less than 0.05, which indicates that the difference between the two models' F1 scores is statistically significant.

Gradient Boosting had a higher mean F1 score (0.6844) than Logistic Regression (0.6732), so Gradient Boosting is preferred.

  Feature Selection and Dimensionality Check

After preprocessing and one-hot encoding, the dataset had **118 features**.

SelectKBest with mutual information was used to select the top 20 features. The results were compared with all 118 features.

| Version              |    Mean F1 | Mean ROC AUC | Mean Fit Time |
| -------------------- | ---------: | -----------: | ------------: |
| All 118 Features     | **0.6732** |   **0.9129** |       0.882 s |
| Selected 20 Features |     0.6517 |       0.9045 |   **0.345 s** |

Using only 20 features reduced the training time, but the F1 score and ROC AUC also decreased.

Therefore, reducing the features did not give better overall results. The 118 features will be retained for the next stage.
