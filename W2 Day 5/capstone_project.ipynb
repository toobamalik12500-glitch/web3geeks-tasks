# Task 1: System Design

## Project: Student Support Agent

The Student Support Agent helps students answer common university-related questions. The agent uses a university information file to find relevant information and gives a simple answer to the student.

### Framework Choice

I will use LangGraph for this project. LangGraph is suitable because the system has multiple steps, such as receiving a student question, checking the information file, generating an answer, and asking for human approval when needed. It also allows us to control the workflow and handle errors. CrewAI is not necessary because this project does not need multiple agents with different roles.


## Architecture Diagram

```text
Student
   ↓
Student Question
   ↓
LangGraph Agent
   ↓
University Information File
   ↓
Generate Answer
   ↓
Student

For important actions:
LangGraph Agent
   ↓
Human Approval
   ↓
Yes → Continue Action
No  → Stop
```

### Main Components

* **Student:** Provides a question or request.
* **LangGraph Agent:** Controls the steps of the system.
* **Information File:** Provides university-related information.
* **Human Approval:** Checks important actions before they are completed.
* **Final Answer:** Gives the result back to the student.

````python

# Read student information from CSV file

import pandas as pd

student_data = pd.read_csv("university_info.csv")

print(student_data)
````

````python
# Task 2 - Step 3
# Function to search student information

def search_student(student_name):
    result = student_data[student_data["Name"].str.lower() == student_name.lower()]
    
    if len(result) == 0:
        return "Student not found."
    
    return result
````

````python
# Test the search function

print(search_student("Ali"))
````

````python
# Task 2 - Step 4
# Example student question

student_name = "Ali"
question = "What is my course?"

print("Student:", student_name)
print("Question:", question)
````

````python
# Task 2 - Step 3
# Function to answer simple student questions

def answer_question(student_name, question):
    result = student_data[student_data["Name"].str.lower() == student_name.lower()]

    if len(result) == 0:
        return "Student not found."

    student = result.iloc[0]

    question = question.lower()

    if "course" in question:
        return f"{student_name} is enrolled in {student['Course']} course."

    elif "teacher" in question:
        return f"The teacher for {student['Course']} is {student['Teacher']}."

    elif "deadline" in question:
        return f"The assignment deadline is {student['Assignment_Deadline']}."

    elif "room" in question:
        return f"The class room is {student['Room']}."

    else:
        return "Sorry, I could not understand your question."
````

````python
# Task 2 - Step 5
# Get answer for the student's question

answer = answer_question(student_name, question)

print("Answer:", answer)
````

````python
# Task 2 - Step 6
# Import LangGraph

from langgraph.graph import StateGraph, START, END

print("LangGraph is ready!")
````

````python
# Task 2 - Step 7
# Create the state for our agent

from typing import TypedDict

class StudentState(TypedDict):
    student_name: str
    question: str
    answer: str

print("State created successfully!")
````

````python
# Task 2 - Step 8
# Create the answer node

def answer_node(state):
    answer = answer_question(
        state["student_name"],
        state["question"]
    )

    return {
        "answer": answer
    }

print("Answer node created!")
````

````python
# Task 2 - Step 9
# Create the LangGraph workflow

graph_builder = StateGraph(StudentState)

graph_builder.add_node("answer", answer_node)

graph_builder.add_edge(START, "answer")
graph_builder.add_edge("answer", END)

student_graph = graph_builder.compile()

print("LangGraph workflow created!")
````

````python
# Task 2 - Step 10
# Run the LangGraph agent

input_data = {
    "student_name": "Ali",
    "question": "What is my course?"
}

result = student_graph.invoke(input_data)

print("Student:", result["student_name"])
print("Question:", result["question"])
print("Answer:", result["answer"])
````

````python
# Task 2 - Step 11
# Test error handling

test_cases = [
    {"student_name": "Ali", "question": "What is my course?"},
    {"student_name": "Unknown", "question": "What is my course?"},
    {"student_name": "Sara", "question": "What is my hobby?"}
]

for test in test_cases:
    result = student_graph.invoke(test)
    
    print("Student:", test["student_name"])
    print("Question:", test["question"])
    print("Answer:", result["answer"])
    print("-" * 40)
````

````python
# Task 2 - Step 12
# Human approval for an important action

action = "Send email to teacher"
approval = "yes"   # Change to "no" to reject the action

print("Requested action:", action)

if approval.lower() == "yes":
    print("Human approved the action.")
    print("Action can continue.")
else:
    print("Human rejected the action.")
    print("Action stopped.")
````

````python
# Task 2 - Step 13
# Create the human approval node

def approval_node(state):
    approval = state.get("approval", "no")

    if approval.lower() == "yes":
        return {
            "answer": state["answer"] + " Human approved the action."
        }
    else:
        return {
            "answer": state["answer"] + " Human rejected the action."
        }

print("Approval node created!")
````

````python
# Task 2 - Step 14
# Connect the approval node to the graph

graph_builder = StateGraph(StudentState)

graph_builder.add_node("answer", answer_node)
graph_builder.add_node("approval", approval_node)

graph_builder.add_edge(START, "answer")
graph_builder.add_edge("answer", "approval")
graph_builder.add_edge("approval", END)

student_graph = graph_builder.compile()

print("Complete LangGraph workflow created!")
````

````python
# Task 2 - Step 15
# Test the complete graph with human approval

input_data = {
    "student_name": "Ali",
    "question": "What is my course?",
    "answer": "",
    "approval": "yes"
}

result = student_graph.invoke(input_data)

print("Student:", result["student_name"])
print("Question:", result["question"])
print("Answer:", result["answer"])
````

````python
# Task 2 - Step 16
# Add approval information to the state

class StudentState(TypedDict):
    student_name: str
    question: str
    answer: str
    approval: str

print("Updated state created successfully!")
````

````python
# Task 2 - Step 17
# Rebuild the LangGraph workflow

graph_builder = StateGraph(StudentState)

graph_builder.add_node("answer", answer_node)
graph_builder.add_node("approval", approval_node)

graph_builder.add_edge(START, "answer")
graph_builder.add_edge("answer", "approval")
graph_builder.add_edge("approval", END)

student_graph = graph_builder.compile()

print("Updated LangGraph workflow created!")
````

````python
# Task 2 - Step 18
# Test the updated graph

input_data = {
    "student_name": "Ali",
    "question": "What is my course?",
    "answer": "",
    "approval": "yes"
}

result = student_graph.invoke(input_data)

print("Student:", result["student_name"])
print("Question:", result["question"])
print("Answer:", result["answer"])
````

````python
# Task 4 - Step 2
# Create a simple FastAPI endpoint

from fastapi import FastAPI

app = FastAPI()

@app.get("/student")
def student_question(student_name: str, question: str):
    answer = answer_question(student_name, question)

    return {
        "student": student_name,
        "question": question,
        "answer": answer
    }

print("FastAPI endpoint created!")
````

````python
# Task 4 - Step 4
# Add basic monitoring

import time

def monitored_student_question(student_name, question):
    start_time = time.time()

    print("Input:", student_name, "-", question)
    print("Function called: answer_question")

    try:
        answer = answer_question(student_name, question)

        end_time = time.time()
        latency = end_time - start_time

        print("Answer:", answer)
        print("Latency:", round(latency, 4), "seconds")
        print("Status: Success")

        return {
            "student": student_name,
            "question": question,
            "answer": answer,
            "latency": round(latency, 4),
            "status": "Success"
        }

    except Exception:
        print("Status: Error")

        return {
            "student": student_name,
            "question": question,
            "answer": "An error occurred.",
            "status": "Error"
        }

print("Monitoring added successfully!")
````

````python
# Task 4 - Step 5
# Test the monitoring function

result = monitored_student_question(
    "Ali",
    "What is my course?"
)

print("Final Result:")
print(result)
````

````python
# Task 4 - Step 6
# Test error handling in monitoring

result = monitored_student_question(
    "Unknown",
    "What is my course?"
)

print("Final Result:")
print(result)
````

- **Token Usage:** Not applicable because this version uses a direct Python function and CSV file instead of an LLM.

## Thresholds

- Latency should normally stay below 1 second.
- Errors should be checked and handled without crashing the system.
- Unexpected answers should be reviewed.

The system should be tested again whenever the CSV data, answer function, or API code is changed.
