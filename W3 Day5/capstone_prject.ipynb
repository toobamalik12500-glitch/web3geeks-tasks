```python
# Task 1 - Load Saved Models

match_winner_model = joblib.load("match_winner_model.pkl")
player_model = joblib.load("top_player_model.pkl")

print("Models loaded successfully")
```

```python
# Task 1 - Create Prediction Functions

def predict_match_winner(team_name, opponent, venue, win_streak, last_5_avg_score, head_to_head_wins):
    data = pd.DataFrame({
        "team_name": [team_name],
        "opponent": [opponent],
        "venue": [venue],
        "win_streak": [win_streak],
        "last_5_avg_score": [last_5_avg_score],
        "head_to_head_wins": [head_to_head_wins]
    })

    prediction = match_winner_model.predict(data)

    return prediction[0]


def predict_player_points(team, opponent, kicks, marks, handballs, disposals, goals, tackles):
    data = pd.DataFrame({
        "team": [team],
        "opponent": [opponent],
        "kicks": [kicks],
        "marks": [marks],
        "handballs": [handballs],
        "disposals": [disposals],
        "goals": [goals],
        "tackles": [tackles]
    })

    prediction = player_model.predict(data)

    return prediction[0]


print("Prediction functions created successfully")
```

```python
# Task 1 - Add Error Handling

def safe_match_prediction(team_name, opponent, venue, win_streak, last_5_avg_score, head_to_head_wins):
    try:
        result = predict_match_winner(
            team_name,
            opponent,
            venue,
            win_streak,
            last_5_avg_score,
            head_to_head_wins
        )

        return result

    except Exception as e:
        return "Unable to make prediction. Please check the input."


def safe_player_prediction(team, opponent, kicks, marks, handballs, disposals, goals, tackles):
    try:
        result = predict_player_points(
            team,
            opponent,
            kicks,
            marks,
            handballs,
            disposals,
            goals,
            tackles
        )

        return result

    except Exception as e:
        return "Unable to make player prediction. Please check the input."


print("Error handling added successfully")
```

```python
# Task 1 - Prompt Injection Check

def check_prompt_injection(query):

    query_lower = query.lower()

    blocked_phrases = [
        "ignore previous instructions",
        "ignore all instructions",
        "forget afl",
        "ignore afl rules"
    ]

    for phrase in blocked_phrases:
        if phrase in query_lower:
            return "Request blocked. I can only answer AFL-related questions."

    afl_keywords = [
        "afl",
        "football",
        "team",
        "player",
        "match",
        "goal",
        "captain",
        "score",
        "ladder",
        "win",
        "winner"
    ]

    for word in afl_keywords:
        if word in query_lower:
            return "Request allowed."

    return "Request blocked. I can only answer AFL-related questions."
```
