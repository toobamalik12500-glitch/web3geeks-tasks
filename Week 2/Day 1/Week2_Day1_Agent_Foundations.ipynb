 Task 1: Agent Concepts & Mental Model

### Agent vs Chatbot vs Workflow

### Agent
An agent is an AI system that can work towards a goal by making decisions and taking different steps. It can use tools when needed and can check the results before deciding what to do next.

### Chatbot
A chatbot is an AI system that mainly talks with users and gives answers to their questions. It usually responds to the user's message without performing many actions by itself.

### Workflow
A workflow is a fixed sequence of steps used to complete a task. The steps are already defined, so the system normally follows the same process every time.

### Main Difference
The main difference is that a chatbot mainly gives answers, a workflow follows fixed steps, while an agent can decide which steps and tools are needed to complete a goal.

## 2. What Makes Something Agentic?

A system is called agentic when it can do more than just give a simple answer. It can make decisions, use tools, perform multiple steps, and adjust its actions based on the results.

### Autonomy
Autonomy means that the agent can decide what action to take next without the user giving every single step.

### Tool Use
An agent can use tools to perform tasks. For example, it can use a calculator to calculate numbers or a weather tool to get weather information.

### Multi-Step Planning
An agent can break a bigger task into smaller steps and complete them one by one.

### Self-Correction
Self-correction means that the agent can check the result of an action and change its next step if something did not work correctly.

These abilities make a system more agentic because it can work towards a goal instead of only answering a question.

## 3. ReAct Pattern

ReAct stands for Reason, Act, and Observe.

It is a pattern where an AI agent thinks about what it needs to do, takes an action using a tool, observes the result, and then decides what to do next.

### ReAct Steps

1. **Reason** – The agent thinks about what step is needed.
2. **Act** – The agent uses a tool or performs an action.
3. **Observe** – The agent receives and checks the result.
4. **Repeat** – The agent reasons again and takes another action if needed.
5. **Final Answer** – When the task is complete, the agent gives the final answer.

### Simple Diagram

Goal
↓
Reason
↓
Act (Use Tool)
↓
Observe Result
↓
Reason Again
↓
Act Again
↓
Observe Again
↓
Final Answer

### Simple Example

If the user asks:

"Check the weather in Lahore and Karachi and tell me which city is warmer."

The agent can:

- Reason: I need the weather of both cities.
- Act: Use the weather tool for Lahore.
- Observe: Lahore is 35°C.
- Act: Use the weather tool for Karachi.
- Observe: Karachi is 31°C.
- Reason: Lahore is warmer.
- Final Answer: Lahore is warmer than Karachi.

## 4. When Is an Agent Overkill?

An agent is not always necessary. If a task is very simple and can be completed with one prompt or a small Python script, using an agent can make the solution more complicated than needed.

For example, simple tasks like converting units, calculating a number, or formatting text can usually be done with a normal function or script. Agents are more useful when a task needs multiple steps, decision-making, tool use, or changing actions based on results.

## Step 1: Define the Tools

In this task, I will use two simple tools:

1. **Calculator Tool**
   This tool will perform basic mathematical calculations.

2. **Weather Lookup Tool**
   This tool will return sample weather information for a given city.

These tools will help the AI model perform actions instead of only giving a text response.


```python
# Calculator Tool

def calculator(a, b, operation):
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        if b == 0:
            return "Cannot divide by zero"
        return a / b
    else:
        return "Invalid operation"
```

```python
calculator(10, 5, "add")
```

```python
# Weather Lookup Tool

def weather_lookup(city):
    weather_data = {
        "Lahore": "35°C, Sunny",
        "Karachi": "31°C, Cloudy",
        "Islamabad": "28°C, Partly Cloudy"
    }

    if city in weather_data:
        return weather_data[city]
    else:
        return "Weather data not available for this city"
```

```python
weather_lookup("Lahore")
```

```python
# Task 3: Build a Minimal Agent Loop
# Step 1: Create a simple agent loop

# The user's question
user_question = "Look up the weather in Lahore and Karachi and tell me which city is warmer."

# Maximum number of times the agent can run
max_iterations = 5

# Start the conversation
messages = [user_question]

# Start the agent loop
for i in range(max_iterations):

    print("\nIteration:", i + 1)

    # Send the question to Gemini
    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=messages,
        config=types.GenerateContentConfig(
            tools=[gemini_tools]
        )
    )

    # Check if Gemini wants to use a tool
    if response.function_calls:

        for call in response.function_calls:

            print("Tool Call:", call.name)
            print("Arguments:", call.args)

            # Run the weather tool
            if call.name == "weather_lookup":

                city = call.args["city"]
                result = weather_lookup(city)

                print("Tool Result:", result)

                # Add the result to the conversation
                messages.append(
                    f"The result of {call.name} for {city} is: {result}"
                )

    else:
        # Gemini has given the final answer
        print("Final Answer:")
        print(response.text)

        break
```

## Task 4: Memory & State Handling

### Conversation Memory

Conversation memory means keeping the previous messages, tool calls, and tool results during a conversation. It helps the agent remember what happened earlier.

### Working Memory

Working memory means temporary information that the agent uses while solving the current task. For example, the temperatures of Lahore and Karachi are temporary information needed to decide which city is warmer.

### Difference

Conversation memory keeps the history of the conversation, while working memory stores temporary information needed to complete the current task.

## Task 5: Failure Modes & Guardrails

### Step 4: Common Failure Modes and Mitigations

| Failure Mode            | Mitigation                                                                                 |
| ----------------------- | ------------------------------------------------------------------------------------------ |
| Infinite loops          | Set a maximum number of iterations so the agent stops after a fixed limit.                 |
| Hallucinated tool calls | Only allow the agent to use tools that are actually defined.                               |
| Wrong tool arguments    | Validate the tool arguments before executing the tool.                                     |
| Tool errors             | Catch errors and return a clear error message instead of crashing.                         |
| Missing information     | Tell the user that the required information is unavailable instead of making up an answer. |
| Wrong tool selection    | Give clear tool descriptions so the agent can choose the correct tool.                     |


### Step 5: Why Agent Frameworks Exist

Agent frameworks such as LangChain, LangGraph, and CrewAI exist to make it easier to build and manage complex AI agents.

* **LangChain:** Helps connect language models with tools, memory, and other components.
* **LangGraph:** Helps build agents with multiple steps, loops, states, and more controlled workflows.
* **CrewAI:** Helps create systems where multiple AI agents work together on different tasks.

For simple agents, raw Python code is enough. However, as an agent becomes more complex, these frameworks can reduce the amount of code and make the system easier to manage.
