## Task 2 - Prediction Targets

### Match Winner
The match winner target has three classes:
- Home Win
- Away Win
- Draw

The target is created from the home team's match result. If the home team result is W, the target is Home Win. If the result is L, the target is Away Win. If the result is D, the target is Draw.

### Top Player by Disposals
The top player is the player with the highest number of disposals in a match for each team.

### Top Player by Goals
The top player is the player with the highest number of goals in a match for each team.

| Target | Definition | Level |
|---|---|---|
| Match Winner | Home Win, Away Win or Draw | Match |
| Top Player by Disposals | Player with highest disposals | Player-Game |
| Top Player by Goals | Player with highest goals | Player-Game |

### Classification or Regression

Match Winner is a classification problem because the target has three categories: Home Win, Away Win and Draw.

We choose classification because we want the model to predict a category, not a continuous number such as match margin.

```python
# Task 4 - Team Win Streak

team_matches["win_streak"] = (
    team_matches.groupby("team_name")["result"]
    .transform(lambda x: x.eq("W").astype(int).groupby((x != "W").cumsum()).cumsum().shift(1).fillna(0))
)

print(team_matches[["team_name", "match_date", "result", "win_streak"]].head(10))
```

```python
# Task 4 - Last 5 Games Average Score

team_matches["last_5_avg_score"] = (
    team_matches.groupby("team_name")["team_score"]
    .transform(lambda x: x.shift(1).rolling(5).mean())
)

print(team_matches[[
    "team_name",
    "match_date",
    "team_score",
    "last_5_avg_score"
]].head(10))
```

```python
# Task 4 - Player Last 5 Games Average Disposals

round_stats["last_5_avg_disposals"] = (
    round_stats.groupby("player_id")["disposals"]
    .shift(1)
    .groupby(round_stats["player_id"])
    .rolling(5)
    .mean()
    .reset_index(level=0, drop=True)
)

print(round_stats[[
    "player_id",
    "match_date",
    "disposals",
    "last_5_avg_disposals"
]].head(10))
```

```python
# Task 4 - Head-to-Head Wins

team_matches["head_to_head_wins"] = 0

print(team_matches[[
    "team_name",
    "opponent",
    "result",
    "head_to_head_wins"
]].head(10))
```

```python
# Task 4 - Head-to-Head Wins

team_matches["head_to_head_wins"] = 0

for i in range(len(team_matches)):
    team = team_matches.iloc[i]["team_name"]
    opponent = team_matches.iloc[i]["opponent"]

    previous = team_matches[
        (team_matches["team_name"] == team) &
        (team_matches["opponent"] == opponent) &
        (team_matches.index < team_matches.index[i])
    ]

    wins = (previous["result"] == "W").sum()

    team_matches.loc[team_matches.index[i], "head_to_head_wins"] = wins

print(team_matches[[
    "team_name",
    "opponent",
    "head_to_head_wins"
]].head(10))
```

```python
# Task 4 - Simple Ladder Position

team_wins = team_matches.groupby("team_name")["result"].apply(
    lambda x: (x == "W").sum()
)

ladder = team_wins.sort_values(ascending=False)

ladder_position = ladder.rank(method="min", ascending=False)

print(ladder_position.sort_values().head(10))
```

```python
# Task 5 - Time Based Split Function

def time_split(data, split_year):
    train = data[data["year"] <= split_year]
    holdout = data[data["year"] > split_year]
    
    return train, holdout
```

## Task 5 - Time-Based Train/Hold-out Split

A time-based split was used instead of a random split. The training data contains matches from 1983 to 2020, while the hold-out data contains matches from 2021 to 2025.

A random split can leak future information because older and newer matches can be mixed between training and testing. A time-based split is better because it uses past matches for training and future matches for testing.

A perfect or very high accuracy would be a warning sign of data leakage. AFL match results are affected by many factors, so a model cannot be expected to predict every match correctly.
