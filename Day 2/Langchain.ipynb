```python
from langchain_core.tools import tool

# Tool to add two numbers
@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers and return their sum."""
    return a + b

# Test the tool
print(add_numbers.invoke({"a": 10, "b": 5}))
```

```python
# Tool to multiply two numbers
@tool
def multiply_numbers(a: int, b: int) -> int:
    """Multiply two numbers and return the result."""
    return a * b

# Test the tool
print(multiply_numbers.invoke({"a": 6, "b": 4}))
```

```python
import pandas as pd

# Create a small product dataset
data = {
    "product": ["Laptop", "Mouse", "Keyboard"],
    "price": [80000, 2500, 4000]
}

df = pd.DataFrame(data)

# Save the data as a CSV file
df.to_csv("products.csv", index=False)

print("CSV file created successfully!")
```

```python
# Tool to get product price from CSV file
@tool
def get_product_price(product_name: str) -> str:
    """Read the products CSV file and return the price of a product."""
    
    df = pd.read_csv("products.csv")
    
    result = df[df["product"].str.lower() == product_name.lower()]
    
    if result.empty:
        return "Product not found."
    
    price = result.iloc[0]["price"]
    return f"The price of {product_name} is {price}."

# Test the tool
print(get_product_price.invoke({"product_name": "Laptop"}))
```

```python
# Put all tools in a list
tools = [
    add_numbers,
    multiply_numbers,
    get_product_price
]

print("All tools registered successfully!")
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
```

```python
# Create a simple prompt for the agent
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Use tools when needed."),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad")
])

print("Agent prompt created successfully!")
```

```python
# Task 3 - Import the required LangChain classes
from langchain.agents import create_agent

print("Import successful!")
```

```python
# Task 3 - Create the agent
agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt="You are a helpful assistant. Use tools when needed."
)

print("Agent created successfully!")
```

```python
# Task 3 - Test the agent with a multi-step question
response = agent.invoke({
    "messages": [
        ("user", "What is the price of the Laptop, and then multiply that price by 2?")
    ]
})

# Task 3 - Print the final answer
print(response["messages"][-1].content)
```

# Task 3 - Comparison with Raw Python Agent

# In the raw Python agent, I manually handled the agent loop,
# tool selection, and tool execution.
# LangChain makes these steps easier by providing an agent framework
# that can automatically call the registered tools.
# The tool calls are visible in the verbose output, while some
# internal reasoning is handled by the framework.

```python
# Task 4 - Import memory and message history tools
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

# Task 4 - Create storage for conversation history
store = {}

# Task 4 - Function to get conversation history
def get_session_history(session_id):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

print("Memory setup successfully!")
```

```python
# Task 4 - Connect the agent with conversation memory
agent_with_memory = RunnableWithMessageHistory(
    agent,
    get_session_history,
    input_messages_key="messages",
    history_messages_key="history"
)

print("Agent memory connected successfully!")
```

```python
# Task 4 - Turn 1: Ask for Laptop price
response1 = agent_with_memory.invoke(
    {"messages": [("user", "What is the price of the Laptop?")]},
    config={"configurable": {"session_id": "student1"}}
)

print(response1["messages"][-1].content)
```

```python
# Task 4 - Turn 2: Ask about another product
response2 = agent_with_memory.invoke(
    {"messages": [("user", "What is the price of the Mouse?")]},
    config={"configurable": {"session_id": "student1"}}
)

print(response2["messages"][-1].content)
```

```python
# Task 4 - Turn 3: Ask for a recommendation using previous context
response3 = agent_with_memory.invoke(
    {"messages": [("user", "Which one is cheaper, the Laptop or the Mouse?")]},
    config={"configurable": {"session_id": "student1"}}
)

print(response3["messages"][-1].content)
```

```python
# Task 5 - Import Pydantic
from pydantic import BaseModel, Field

# Task 5 - Create a simple output structure
class ProductAnswer(BaseModel):
    product: str = Field(description="Product name")
    price: float = Field(description="Product price")
    recommendation: str = Field(description="Simple recommendation")

print("Structured output model created successfully!")
```

```python
# Task 5 - Connect Gemini with the structured output
structured_llm = llm.with_structured_output(ProductAnswer)

print("Structured output connected successfully!")
```

```python
# Task 5 - Test the structured output
response = structured_llm.invoke(
    "Give the price and recommendation for a Laptop. The price is 80000."
)

# Task 5 - Print the structured answer
print(response)
```

```python
# Task 5 - Create a tool that can give an error
@tool
def divide_numbers(a: int, b: int) -> str:
    """Divide one number by another number."""
    
    try:
        result = a / b
        return f"The answer is {result}"
    
    except ZeroDivisionError:
        return "Error: Cannot divide by zero."

# Task 5 - Test the tool
print(divide_numbers.invoke({"a": 10, "b": 2}))
print(divide_numbers.invoke({"a": 10, "b": 0}))
```
