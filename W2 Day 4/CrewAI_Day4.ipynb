# Task 1 — Multi-Agent Design

## Business Problem

A small online clothing store wants to launch a new product. The store needs to understand customer interests, create a product description, and check the final content before publishing it.

## Agent 1 — Customer Researcher

Role: Customer Researcher

Goal: Find what type of clothing customers may like and what features are important to them.

Backstory: You are a customer researcher who studies customer interests and simple market trends. You focus on finding useful information for the clothing store.

## Agent 2  Product Content Writer

Role: Product Content Writer

Goal: Use the research to write a clear and attractive product description for the new clothing product.

Backstory: You are a creative content writer who writes simple and attractive product descriptions for online stores. You use customer information to make the content useful for customers.

## Agent 3 Content Reviewer

Role: Content Reviewer

Goal: Check the product description for mistakes, missing information, and clarity, and suggest improvements.

Backstory: You are a careful reviewer who checks online store content before it is published. You focus on making sure the content is clear, complete, and suitable for customers.

## Why Multiple Specialized Agents?

Multiple agents can be useful because each agent focuses on one specific job. The researcher focuses on customer information, the writer creates the product description, and the reviewer checks the final result. However, for a simple task, one general agent may be enough and using multiple agents can add extra complexity.

```python
# Task 3 - Create Customer Research Task

from crewai import Task

research_task = Task(
    description="Research customer interests for a new clothing product. Identify popular clothing styles, useful features, and what customers may look for when buying online.",
    expected_output="A short list of customer interests, popular clothing styles, and important product features.",
    agent=researcher
)

print("Research task created successfully!")
```

```python
# Task 3 - Create Product Description Task

writing_task = Task(
    description="Use the customer research from the previous task to write a clear and attractive product description for a new clothing product.",
    expected_output="A short product description including the product name, main features, benefits, and why customers may like it.",
    agent=writer,
    context=[research_task]
)

print("Writing task created successfully!")
```

```python
# Task 3 - Create Content Review Task

review_task = Task(
    description="Review the product description created in the previous task. Check for mistakes, missing information, clarity, and suitability for online customers. Suggest improvements where needed.",
    expected_output="A reviewed product description with clear suggestions for improvement and a final version ready for publishing.",
    agent=reviewer,
    context=[writing_task]
)

print("Review task created successfully!")
```

```python
# Task 3 - Create Sequential Crew

from crewai import Crew, Process

crew = Crew(
    agents=[researcher, writer, reviewer],
    tasks=[research_task, writing_task, review_task],
    process=Process.sequential,
    verbose=True
)

print("Sequential crew created successfully!")
```

```python
# Task 3 - Run the Sequential Crew asynchronously

result = await crew.kickoff_async(
    inputs={
        "product": "New summer cotton dress"
    }
)

print(result)
```

```python
# Task 4 - Create Manager Agent

from crewai import Agent

manager_agent = Agent(
    role="Crew Manager",
    goal="Manage the team and make sure the final product content is useful and complete.",
    backstory="You are an experienced team manager who delegates tasks to specialized agents and reviews their work.",
    llm=llm,
    allow_delegation=True,
    verbose=True
)

print("Manager Agent is ready!")
```

```python
# Task 4 - Create Hierarchical Crew

from crewai import Crew, Process

hierarchical_crew = Crew(
    agents=[
        researcher,
        writer,
        reviewer
    ],
    tasks=[
        research_task,
        writing_task,
        review_task
    ],
    manager_agent=manager_agent,
    process=Process.hierarchical,
    verbose=True
)

print("Hierarchical Crew is ready!")
```

```python
# Task 4 - Run the Hierarchical Crew asynchronously

result_hierarchical = await hierarchical_crew.kickoff_async(
    inputs={
        "product": "New summer cotton dress"
    }
)

print(result_hierarchical)
```

## Task 4 - Sequential vs Hierarchical Comparison

| Aspect | Sequential Process | Hierarchical Process |
|---|---|---|
| Process | Agents work in a fixed order | Manager delegates and reviews the work |
| Quality | Good and consistent | More flexible and detailed |
| Latency/Cost | Lower because fewer LLM calls are needed | Higher because the manager may make extra LLM calls |
| Reliability | Simple and predictable | More flexible but depends on manager decisions |
| Pros | Simple, easy to understand, predictable | Better delegation, review, and flexibility |
| Cons | Less flexible and no manager supervision | More complex and can cost more |
| When to Use | When tasks have a clear fixed order | When tasks need delegation, supervision, or review |

## Task 5 - Success Criteria

We will evaluate the crew's output using three simple success criteria:

1. **Factual Grounding:** The product description should be based on the research and task information.

2. **Completeness:** The product description should include important details such as features, fit, colors, and care instructions.

3. **Clear Tone:** The description should be clear, attractive, and easy for customers to understand.

### For this specific task, was a multi-agent crew worth the added complexity and cost compared to a single well-designed agent?

For this specific task, the multi-agent crew was not worth the added complexity and cost. The sequential approach had a lower estimated cost of $0.0803, while the hierarchical approach cost $0.3179. A single well-designed agent could handle this task more simply and at a lower cost. The multi-agent approach may be more useful for complex tasks that require multiple specialized agents.
