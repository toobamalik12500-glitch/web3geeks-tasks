# Week 4 - Day 1
## Task 1 - Modern Voice Agent Architecture

A real estate AI voice agent needs several components to understand the customer, retrieve correct property information, generate a response, and speak naturally.

### Main Components

1. **Speech-to-Text (STT)**  
   Converts the customer's spoken voice into text.

2. **LLM (Large Language Model)**  
   Understands the customer's question and generates a natural response.

3. **Tool Calling**  
   Allows the agent to use tools such as property search, Google Calendar, email, and CRM.

4. **RAG (Retrieval-Augmented Generation)**  
   Retrieves relevant information from the company's property knowledge base before generating an answer.

5. **Memory**  
   Stores important conversation information such as budget, city, property requirements, and previous requests.

6. **Text-to-Speech (TTS)**  
   Converts the AI-generated text into natural spoken voice.

7. **Telephony**  
   Connects the AI agent with incoming and outgoing phone calls.

8. **Workflow Orchestration**  
   Coordinates different steps such as property search, appointment booking, Calendar updates, email notifications, and CRM logging.

### Basic Architecture Flow

Customer Voice  
↓  
Speech-to-Text  
↓  
LLM / Intent Detection  
↓  
RAG + Tools + Memory  
↓  
LLM Response  
↓  
Text-to-Speech  
↓  
Customer Voice

### Conclusion

The architecture combines voice processing, AI reasoning, retrieval, memory, tool calling, and business workflows to create a conversational real estate voice agent.

```python
# Task 27 - Final Zameen Dataset
# 50 unique properties from each city
# One combined CSV file

import requests
from bs4 import BeautifulSoup
import pandas as pd
import re
import time

# Cities
cities = {
    "Faisalabad": "https://www.zameen.com/Rentals_Houses_Property/Faisalabad-16-",
    "Lahore": "https://www.zameen.com/Rentals_Houses_Property/Lahore-1-",
    "Islamabad": "https://www.zameen.com/Rentals_Houses_Property/Islamabad-3-",
    "Karachi": "https://www.zameen.com/Rentals_Houses_Property/Karachi-2-"
}

# Browser header
headers = {
    "User-Agent": "Mozilla/5.0"
}

# Store all final properties
all_properties = []


# Function to extract property ID from URL
def get_property_id(url):

    match = re.search(
        r"-(\d+)-\d+-\d+\.html",
        url
    )

    if match:
        return match.group(1)

    return None


# Function to extract amenities
def extract_amenities(page_text):

    if "Amenities" not in page_text:
        return "Not available"

    amenities_text = page_text.split(
        "Amenities",
        1
    )[1]

    if "Location & Nearby" in amenities_text:

        amenities_text = amenities_text.split(
            "Location & Nearby",
            1
        )[0]

    amenities_text = re.sub(
        r"\s+",
        " ",
        amenities_text
    ).strip()

    amenities_text = re.sub(
        r"^Main Features\s*",
        "",
        amenities_text,
        flags=re.IGNORECASE
    )

    if amenities_text:
        return amenities_text

    return "Not available"


# Start city by city
for city, base_url in cities.items():

    print("\nStarting:", city)

    # Store unique property IDs
    collected_ids = set()

    # Store unique property URLs
    property_links = []

    # Start from page 1
    page_number = 1

    # Keep checking pages until 50 UNIQUE properties
    while len(collected_ids) < 50:

        listing_url = (
            base_url
            + str(page_number)
            + ".html"
        )

        print(
            "Checking",
            city,
            "page:",
            page_number
        )

        try:

            # Request listing page
            response = requests.get(
                listing_url,
                headers=headers,
                timeout=30
            )

            print(
                "Status code:",
                response.status_code
            )

            # Stop if page failed
            if response.status_code != 200:

                print(
                    "Page could not be accessed."
                )

                break

            # Read HTML
            soup = BeautifulSoup(
                response.text,
                "html.parser"
            )

            # Find property links
            for link in soup.find_all(
                "a",
                href=True
            ):

                href = link["href"]

                # Ignore email links
                if href.startswith("mailto:"):
                    continue

                # Keep only property pages
                if "/Property/" not in href:
                    continue

                # Convert relative URL
                if href.startswith("/"):
                    href = (
                        "https://www.zameen.com"
                        + href
                    )

                # Extract property ID
                property_id = get_property_id(
                    href
                )

                # Ignore links without ID
                if property_id is None:
                    continue

                # Only collect NEW property IDs
                if property_id not in collected_ids:

                    collected_ids.add(
                        property_id
                    )

                    property_links.append(
                        href
                    )

                # Stop after 50 unique IDs
                if len(collected_ids) >= 50:
                    break

            print(
                "Unique properties collected:",
                len(collected_ids)
            )

            # Move to next page
            page_number += 1

            # Small delay
            time.sleep(2)

        except Exception as e:

            print(
                "Listing page error:",
                e
            )

            break

    # Keep exactly maximum 50
    property_links = property_links[:50]

    print(
        "\nUnique links collected for",
        city,
        ":",
        len(property_links)
    )


    # Open each property page
    for i, property_url in enumerate(
        property_links,
        start=1
    ):

        try:

            # Open property page
            response = requests.get(
                property_url,
                headers=headers,
                timeout=30
            )

            # Skip failed page
            if response.status_code != 200:

                print(
                    city,
                    "- Property",
                    i,
                    "skipped"
                )

                continue

            # Read HTML
            soup = BeautifulSoup(
                response.text,
                "html.parser"
            )

            # Get visible page text
            page_text = soup.get_text(
                " ",
                strip=True
            )

            # Get property ID
            property_id = get_property_id(
                property_url
            )

            if property_id is None:

                property_id = "Not available"


            # Get title
            if soup.title:

                title = soup.title.get_text(
                    strip=True
                )

            else:

                title = "Not available"


            # Get price
            price_match = re.search(
                r"PKR\s*[\d,]+(?:\.\d+)?\s*"
                r"(?:Lakh|Crore|Arab|Thousand)?",
                page_text,
                re.IGNORECASE
            )

            if price_match:

                price = price_match.group(
                    0
                ).strip()

            else:

                price = "Not available"


            # Get bedrooms
            bedroom_match = re.search(
                r"(\d+)\s*Beds?",
                page_text,
                re.IGNORECASE
            )

            if bedroom_match:

                bedrooms = (
                    bedroom_match.group(1)
                )

            else:

                bedrooms = "Not available"


            # Get bathrooms
            bathroom_match = re.search(
                r"(\d+)\s*Baths?",
                page_text,
                re.IGNORECASE
            )

            if bathroom_match:

                bathrooms = (
                    bathroom_match.group(1)
                )

            else:

                bathrooms = "Not available"


            # Get area
            area_match = re.search(
                r"(\d+(?:\.\d+)?)\s*"
                r"(Marla|Kanal|Sq\.?\s*Ft)",
                page_text,
                re.IGNORECASE
            )

            if area_match:

                area = (
                    area_match.group(1)
                    + " "
                    + area_match.group(2)
                )

            else:

                area = "Not available"


            # Get agent name
            agent_match = re.search(
                r"Agent\s*:\s*([A-Za-z ]+)",
                page_text
            )

            if agent_match:

                agent_name = (
                    agent_match.group(1)
                    .strip()
                )

            else:

                agent_name = "Not available"


            # Get amenities
            amenities = extract_amenities(
                page_text
            )


            # Create property record
            property_data = {

                "property_id": property_id,

                "city": city,

                "property_type": "House",

                "title": title,

                "price": price,

                "bedrooms": bedrooms,

                "bathrooms": bathrooms,

                "area": area,

                "description": page_text[:1000],

                "agent_name": agent_name,

                "amenities": amenities,

                "availability": "For Rent"
            }


            # Add property
            all_properties.append(
                property_data
            )

            print(
                city,
                "- Property",
                i,
                "done"
            )

            # Small delay
            time.sleep(1)

        except Exception as e:

            print(
                city,
                "- Property",
                i,
                "error:",
                e
            )


# Convert data to DataFrame
final_df = pd.DataFrame(
    all_properties
)


# Remove duplicate property IDs
if not final_df.empty:

    final_df = final_df.drop_duplicates(
        subset="property_id",
        keep="first"
    )


# Save ONE combined CSV
final_df.to_csv(
    "zameen_properties_all.csv",
    index=False
)


# Final result
print("\nScraping completed.")

print(
    "Total properties:",
    len(final_df)
)

print(
    "\nProperties by city:"
)

print(
    final_df["city"].value_counts()
)

print(
    "\nUnique property IDs:",
    final_df["property_id"].nunique()
)

print(
    "\nONE CSV created:",
    "zameen_properties_all.csv"
)
```

```python
# Task 6 - Convert price into numeric PKR

def convert_price(price):
    # Convert price to string
    price = str(price).strip()

    # Check for Lakh
    if "Lakh" in price:
        number = float(price.replace("Lakh", "").strip())
        return number * 100000

    # Check for Crore
    elif "Crore" in price:
        number = float(price.replace("Crore", "").strip())
        return number * 10000000

    # Check for Thousand
    elif "Thousand" in price:
        number = float(price.replace("Thousand", "").strip())
        return number * 1000

    # If price is already a number
    else:
        try:
            return float(price.replace(",", ""))
        except:
            return None


# Apply the function to price column
final_df["price"] = final_df["price"].apply(convert_price)

# Show cleaned prices
print(final_df["price"].head(10))
```

```python
# Task 7 - Clean bedrooms, bathrooms and area

# Convert bedrooms to numeric
final_df["bedrooms"] = pd.to_numeric(
    final_df["bedrooms"], errors="coerce"
)

# Convert bathrooms to numeric
final_df["bathrooms"] = pd.to_numeric(
    final_df["bathrooms"], errors="coerce"
)

# Clean area text
final_df["area"] = final_df["area"].astype(str).str.strip()

# Show the cleaned columns
print(final_df[["bedrooms", "bathrooms", "area"]].head(10))
```

```python
# Task 2 - Step 2: Create complete property and FAQ documents

# Create an empty list for property documents
property_documents = []

# Convert every property row into a complete text document
for index, row in properties_df.iterrows():

    document = f"""
Property Information

Property ID: {row['property_id']}
City: {row['city']}
Property Type: {row['property_type']}
Title: {row['title']}
Price: {row['price']}
Bedrooms: {row['bedrooms']}
Bathrooms: {row['bathrooms']}
Area: {row['area']}
Description: {row['description']}
Agent Name: {row['agent_name']}
Amenities: {row['amenities']}
Availability: {row['availability']}
"""

    # Add complete property document to the list
    property_documents.append(document)


# Create an empty list for FAQ documents
faq_documents = []

# Convert every FAQ row into a text document
for index, row in faq_df.iterrows():

    document = f"""
FAQ Information

FAQ ID: {row['faq_id']}
Category: {row['category']}
Question: {row['question']}
"""

    # Add FAQ document to the list
    faq_documents.append(document)


# Check document counts
print("Total property documents:", len(property_documents))
print("Total FAQ documents:", len(faq_documents))

# Display the first property document
print("\nFirst property document:")
print(property_documents[0])

# Display the first FAQ document
print("\nFirst FAQ document:")
print(faq_documents[0])
```

```python
# Task 2 - Step 3: Split documents into smaller chunks

# Function to split one document into smaller chunks
def split_document(text, chunk_size=1000, overlap=100):

    # Create an empty list for chunks
    chunks = []

    # Start position of the first chunk
    start = 0

    # Continue until the complete document is processed
    while start < len(text):

        # Find the end position of the current chunk
        end = start + chunk_size

        # Get the current chunk
        chunk = text[start:end]

        # Add the chunk to the list
        chunks.append(chunk)

        # Move forward while keeping some text as overlap
        start = end - overlap

    # Return all chunks
    return chunks


# Create an empty list for all property chunks
property_chunks = []

# Split every property document into chunks
for document in property_documents:

    # Get chunks for the current property
    chunks = split_document(document)

    # Add all chunks to the main list
    property_chunks.extend(chunks)


# Create an empty list for all FAQ chunks
faq_chunks = []

# Split every FAQ document into chunks
for document in faq_documents:

    # Get chunks for the current FAQ
    chunks = split_document(document)

    # Add all chunks to the main list
    faq_chunks.extend(chunks)


# Show the total number of chunks
print("Total property chunks:", len(property_chunks))
print("Total FAQ chunks:", len(faq_chunks))

# Show the first property chunk
print("\nFirst property chunk:")
print(property_chunks[0])

# Show the first FAQ chunk
print("\nFirst FAQ chunk:")
print(faq_chunks[0])
```

```python
# Task 2 - Step 6: Create retriever function

# Function to retrieve relevant chunks for a question
def retrieve_chunks(question, number_of_results=5):

    # Convert the user question into a TF-IDF vector
    question_embedding = vectorizer.transform([question])

    # Search for the most relevant chunks
    distances, indexes = vector_store.kneighbors(
        question_embedding,
        n_neighbors=number_of_results
    )

    # Create an empty list for retrieved chunks
    retrieved_chunks = []

    # Get each matching chunk
    for index in indexes[0]:

        # Add the matching chunk to the list
        retrieved_chunks.append(all_chunks[index])

    # Return the relevant chunks
    return retrieved_chunks


# Test the retriever with a sample question
question = "Kya is property mein gym hai?"

# Retrieve relevant chunks
results = retrieve_chunks(question)

# Show the retrieved results
print("Question:", question)

print("\nRetrieved chunks:")

for i, result in enumerate(results, start=1):

    print("\nResult", i)
    print(result[:500])
```

Task 3 - Structured vs Semantic Retrieval

Structured retrieval is used for price, availability, area, and agent names because these values are stored in organized columns and can be retrieved directly using SQL.

Semantic retrieval is used for property descriptions and FAQs because they contain text and users may ask questions in different words.

Brochures were not included because actual brochure data is not available in our dataset.

This approach helps the system use exact database retrieval for structured property information and semantic retrieval for text-based information.

```python
# Task 4 - Step 1: Create Basic Property Recommendation

def recommend_properties(city, max_budget, area=None, bedrooms=None, amenity=None):

    # Filter properties by city
    results = properties_df[
        properties_df["city"].str.lower() == city.lower()
    ]

    # Filter properties by maximum budget
    results = results[
        results["price"] <= max_budget
    ]

    # Apply area filter if provided
    if area is not None:
        results = results[
            results["area"].str.contains(
                area,
                case=False,
                na=False
            )
        ]

    # Apply bedroom filter if provided
    if bedrooms is not None:
        results = results[
            results["bedrooms"] == bedrooms
        ]

    # Apply amenity filter if provided
    if amenity is not None:
        results = results[
            results["amenities"].str.contains(
                amenity,
                case=False,
                na=False
            )
        ]

    # Return selected property information
    return results[
        [
            "property_id",
            "city",
            "title",
            "price",
            "bedrooms",
            "area",
            "amenities",
            "agent_name"
        ]
    ]


# Test the recommendation function
recommendations = recommend_properties(
    "Faisalabad",
    150000,
    area="10 Marla"
)

print("Recommended properties:")
print(recommendations.head(10))
```

```python
# Task 5 - Step 8: Set Grounding and Missing Information Rule

# Function to check whether retrieved information can answer the question
def check_information_available(question, retrieved_chunk):

    # Convert text to lowercase
    question = question.lower()
    retrieved_chunk = retrieved_chunk.lower()

    # Check for city information
    cities = [
        "faisalabad",
        "lahore",
        "islamabad",
        "karachi"
    ]

    for city in cities:

        if city in question:

            if city not in retrieved_chunk:
                return False

            break

    # Check important information requested by the customer
    if "parking" in question:

        if "parking" not in retrieved_chunk:
            return False

    if "bedroom" in question:

        if "bedroom" not in retrieved_chunk:
            return False

    if "price" in question:

        if "price" not in retrieved_chunk:
            return False

    if "area" in question:

        if "area" not in retrieved_chunk:
            return False

    if "amenities" in question:

        if "amenities" not in retrieved_chunk:
            return False

    # Information is available
    return True


# Test with information that exists
test_question_1 = "Faisalabad mein parking wali property hai?"

test_result_1 = check_information_available(
    test_question_1,
    evaluation_results[0]["retrieved_chunk"]
)

print("Test 1 - Information available:", test_result_1)


# Test with information that may not exist
test_question_2 = "Faisalabad mein swimming pool wali property hai?"

test_result_2 = check_information_available(
    test_question_2,
    evaluation_results[0]["retrieved_chunk"]
)

print("Test 2 - Swimming pool information available:", test_result_2)


# Expected response rule
if test_result_2 == False:

    print(
        "Agent response: Sorry, hamare available dataset mein "
        "swimming pool ke bare mein information nahi hai."
    )
```

```python
# Task 5 - Step 11: Correct Retrieval Accuracy

# Create a dictionary to store one result for each question
retrieval_status = {}

# Check SQL results
for item in sql_results:

    question = item["question"]
    result = item["result"]

    # SQL retrieval is successful if database returned data
    if len(result) > 0:
        retrieval_status[question] = True
    else:
        retrieval_status[question] = False


# Check semantic results
for item in evaluation_results:

    question = item["question"]
    retrieved_chunk = item["retrieved_chunk"]

    # Only add semantic result if the question
    # has not already been checked through SQL
    if question not in retrieval_status:

        if retrieved_chunk:
            retrieval_status[question] = True
        else:
            retrieval_status[question] = False


# Count successful unique questions
successful_retrievals = sum(
    retrieval_status.values()
)

# Total questions
total_questions = len(evaluation_questions)

# Calculate retrieval accuracy
retrieval_accuracy = (
    successful_retrievals / total_questions
) * 100


print("Total evaluation questions:", total_questions)
print("Successful retrievals:", successful_retrievals)
print(
    "Retrieval Accuracy:",
    round(retrieval_accuracy, 2),
    "%"
)
```

```python
# Task 5 - Step 12: Calculate Grounding Rate

# Count grounded questions
grounded_questions = 0

# Check SQL results
for item in sql_results:

    # SQL result comes directly from the property database
    if len(item["result"]) > 0:
        grounded_questions += 1


# Check semantic results
for item in evaluation_results:

    # Get retrieved chunk
    retrieved_chunk = item["retrieved_chunk"]

    # Semantic result is grounded when
    # actual property information is retrieved
    if retrieved_chunk:
        grounded_questions += 1


# Avoid counting the same questions twice
grounded_questions = min(
    grounded_questions,
    total_questions
)


# Calculate grounding rate
grounding_rate = (
    grounded_questions / total_questions
) * 100


print("Total evaluation questions:", total_questions)
print("Grounded questions:", grounded_questions)
print(
    "Grounding Rate:",
    round(grounding_rate, 2),
    "%"
)
```

```python
## Task 5 - Step 12: Proper Grounding Evaluation

# Create a dictionary for grounding status
grounding_status = {}

# Check SQL results
for item in sql_results:

    question = item["question"]
    result = item["result"]

    # SQL results come directly from the actual database
    if len(result) > 0:
        grounding_status[question] = True
    else:
        grounding_status[question] = False


# Check semantic results
for item in evaluation_results:

    question = item["question"]
    retrieved_chunk = item["retrieved_chunk"]

    # Only evaluate semantic questions here
    if question not in grounding_status:

        # Check whether the retrieved chunk contains
        # information related to the question
        question_words = question.lower().split()

        matched_words = 0

        for word in question_words:

            if len(word) > 3 and word in retrieved_chunk.lower():
                matched_words += 1

        # Consider the result grounded when
        # at least two meaningful words are matched
        if matched_words >= 2:
            grounding_status[question] = True
        else:
            grounding_status[question] = False


# Count grounded questions
grounded_questions = sum(
    grounding_status.values()
)


# Total questions
total_questions = len(evaluation_questions)


# Calculate grounding rate
grounding_rate = (
    grounded_questions / total_questions
) * 100


print("Total evaluation questions:", total_questions)
print("Grounded questions:", grounded_questions)
print(
    "Grounding Rate:",
    round(grounding_rate, 2),
    "%"
)
```

```python
# Task 5 - Step 13: Calculate Hallucination Rate

# Count questions where information was not properly grounded
hallucinated_questions = 0

# Check every evaluation question
for question in evaluation_questions:

    # Get grounding status
    grounded = grounding_status.get(
        question,
        False
    )

    # If information is not grounded,
    # count it as a hallucination case
    if not grounded:
        hallucinated_questions += 1


# Calculate hallucination rate
hallucination_rate = (
    hallucinated_questions / total_questions
) * 100


print("Total evaluation questions:", total_questions)
print(
    "Hallucinated questions:",
    hallucinated_questions
)

print(
    "Hallucination Rate:",
    round(hallucination_rate, 2),
    "%"
)
```

```python
# Week 4 - Day 3
# Task 4 - Improve Objection Recognition

def identify_objection(user_message):

    message = user_message.lower()

    if (
        "expensive" in message
        or "mehnga" in message
        or "mahanga" in message
        or "mahnga" in message
    ):
        return "price"

    elif "trust" in message or "bharosa" in message:
        return "trust"

    elif "location" in message or "jagah" in message:
        return "location"

    elif "investment" in message or "invest" in message:
        return "investment"

    elif "builder" in message:
        return "builder"

    elif "maintenance" in message or "maintain" in message:
        return "maintenance"

    else:
        return "other"
```

```python
# Week 4 - Day 3
# Task 4 - Step 2: Connect Objection Handling with Actual Agent

def handle_objection(user_message):

    # Identify customer's objection
    objection = identify_objection(user_message)

    # Get current conversation memory
    memory_context = get_memory_context()

    # Get city and area from memory
    city = conversation_memory["city"]
    area = conversation_memory["area"]

    # Search actual property data
    if city is not None and area is not None:

        property_results = search_properties(
            city,
            area
        )

    else:

        property_results = pd.DataFrame()

    # Check if matching properties exist
    if len(property_results) == 0:

        print("Agent:")
        print(
            "Ji bilkul, aap ki concern samajh raha hoon. "
            "Lekin is requirement ke mutabiq "
            "hamare available dataset mein property information nahi hai."
        )

        return

    # Convert actual properties into text
    property_context = property_results.to_string(
        index=False
    )

    # Create grounded objection-handling prompt
    prompt = f"""
You are a professional Pakistani real estate voice agent.

Respond in natural UrduLish.

Customer Objection:
{objection}

Customer Message:
{user_message}

Previous Conversation Memory:
{memory_context}

Available Property Data:
{property_context}

Handle the customer's objection politely and naturally.

Use ONLY the property information provided above.

Do not invent:
- prices
- locations
- bedrooms
- bathrooms
- amenities
- agent names
- builder information
- maintenance charges
- investment returns
- any other property information

If the objection is about information that is not available
in the dataset, clearly say that the information is not available.

Do not argue with the customer.
Acknowledge the concern and provide factual information
when available.
"""

    # Send request to Gemini
    response = client.interactions.create(
        model="gemini-3.8-flash",
        input=prompt
    )

    # Get agent response
    agent_answer = response.output_text

    print("Agent:")
    print(agent_answer)


# Test actual objection handling
handle_objection(
    "Ye house bohat mehnga hai"
)
```

```python
# Week 4 - Day 3
# Task 4 - Step 6: Local Objection Handling

def local_objection_agent(user_message):

    # Identify customer objection
    objection = identify_objection(user_message)

    # Get city and area from memory
    city = conversation_memory["city"]
    area = conversation_memory["area"]

    # Search actual property dataset
    if city is not None and area is not None:

        property_results = search_properties(
            city,
            area
        )

    else:

        property_results = pd.DataFrame()

    # Check if properties are available
    if len(property_results) == 0:

        return (
            "Ji bilkul, aap ki concern samajh raha hoon. "
            "Is requirement ke mutabiq hamare dataset mein "
            "matching property available nahi hai."
        )

    # Handle price objection
    if objection == "price":

        lowest_price = property_results["price"].min()

        return (
            "Hmm... ji bilkul, price concern samajh sakta hoon. "
            f"Hamare available options mein lowest listed price "
            f"PKR {lowest_price} hai. "
            "Agar aap chahein to main available options compare kar sakta hoon."
        )

    # Handle location objection
    elif objection == "location":

        return (
            "Ji bilkul, location important hoti hai. "
            "Hamare dataset mein available property "
            f"{city} mein {area} size ke sath listed hai. "
            "Agar aap kisi specific location ki requirement batayein "
            "to main available data check kar sakta hoon."
        )

    # Handle trust objection
    elif objection == "trust":

        return (
            "Acha ji, trust bilkul important hai. "
            "Hamare dataset mein property ke sath agent name available hai, "
            "lekin verification information available nahi hai. "
            "Main sirf available property information share kar sakta hoon."
        )

    # Handle investment objection
    elif objection == "investment":

        return (
            "Ji bilkul, investment ke hawale se main sirf "
            "hamare available property data ki information de sakta hoon. "
            "Investment return ya future profit ka data hamare dataset mein available nahi hai."
        )

    # Handle builder objection
    elif objection == "builder":

        return (
            "Ji, builder information hamare current dataset mein available nahi hai. "
            "Main aap ko property ki available details aur agent information bata sakta hoon."
        )

    # Handle maintenance objection
    elif objection == "maintenance":

        return (
            "Ji bilkul, maintenance charges ki information "
            "hamare current dataset mein available nahi hai."
        )

    # Other questions
    else:

        return (
            "Ji bilkul, aap ki concern samajh raha hoon. "
            "Main available property information ke basis par aap ki help karta hoon."
        )


# Test the local objection agent
print("Agent:")
print(
    local_objection_agent(
        "Ye house bohat mehnga hai"
    )
)
```

```python
# Week 4 - Day 5
# Task 2 - Step 3: Add Conditional Routing

def route_intent(state):

    intent = state["intent"]

    if intent == "property_search":
        return "rag"

    elif intent == "appointment_booking":
        return "booking"

    elif intent == "rescheduling":
        return "rescheduling"

    elif intent == "cancellation":
        return "cancellation"

    else:
        return "goodbye"


graph = StateGraph(AgentState)

graph.add_node("greeting", greeting_node)
graph.add_node("intent_detection", intent_detection_node)
graph.add_node("rag", rag_node)
graph.add_node("recommendation", recommendation_node)
graph.add_node("booking", booking_node)
graph.add_node("rescheduling", rescheduling_node)
graph.add_node("cancellation", cancellation_node)
graph.add_node("email", email_node)
graph.add_node("goodbye", goodbye_node)

graph.add_edge(START, "greeting")
graph.add_edge("greeting", "intent_detection")

graph.add_conditional_edges(
    "intent_detection",
    route_intent,
    {
        "rag": "rag",
        "booking": "booking",
        "rescheduling": "rescheduling",
        "cancellation": "cancellation",
        "goodbye": "goodbye"
    }
)

graph.add_edge("rag", "recommendation")
graph.add_edge("recommendation", "booking")

graph.add_edge("booking", "email")
graph.add_edge("rescheduling", "email")
graph.add_edge("cancellation", "email")

graph.add_edge("email", "goodbye")
graph.add_edge("goodbye", END)

agent_graph = graph.compile()

print("Conditional AI Agent Graph created successfully!")
```

```python
# Week 4 - Day 5
# Task 4 - Step 1: Validate Property Results

def validate_property_results(results):

    if len(results) == 0:

        return {
            "valid": False,
            "message": (
                "Ji, is requirement ke mutabiq "
                "hamare dataset mein koi property available nahi hai."
            )
        }

    return {
        "valid": True,
        "message": "Valid properties found."
    }


print("Property validation created successfully!")
```

```python
# Week 4 - Day 5
# Task 4 - Step 3: Ask for Clarification

def ask_for_clarification(city=None, area=None, appointment_date=None, appointment_time=None):

    missing_information = []

    if city is None:
        missing_information.append("city")

    if area is None:
        missing_information.append("area")

    if appointment_date is None:
        missing_information.append("appointment date")

    if appointment_time is None:
        missing_information.append("appointment time")

    if len(missing_information) > 0:

        return {
            "valid": False,
            "message": (
                "Ji, please thori si aur information dein: "
                + ", ".join(missing_information)
            )
        }

    return {
        "valid": True,
        "message": "Required information is available."
    }


print("Clarification check created successfully!")
```

```python
# Week 4 - Day 5
# Task 4 - Step 4: Combine Validation Checks

def validate_booking(
    city,
    area,
    appointment_date,
    appointment_time,
    property_results
):

    property_check = validate_property_results(
        property_results
    )

    if not property_check["valid"]:
        return property_check

    clarification_check = ask_for_clarification(
        city,
        area,
        appointment_date,
        appointment_time
    )

    if not clarification_check["valid"]:
        return clarification_check

    appointment_check = validate_appointment(
        appointment_date,
        appointment_time
    )

    if not appointment_check["valid"]:
        return appointment_check

    return {
        "valid": True,
        "message": (
            "Property and appointment details "
            "are valid."
        )
    }


print("Booking validation created successfully!")
```

```python
# Week 4 - Day 5
# Task 5 - Step 2: Create Node Transition Logger

def log_transition(state, node_name, message):

    state["execution_trace"].append({
        "node": node_name,
        "message": message
    })

    return state


print("Node transition logger created successfully!")
```

```python
# Week 4 - Day 5
# Task 5 - Step 3: Add Logging to Greeting Node

def greeting_node(state):

    state["conversation_history"].append(
        "Agent: Assalam-o-Alaikum, main aap ki property search mein help kar sakta hoon."
    )

    state = log_transition(
        state,
        "greeting",
        "Greeting message sent to customer."
    )

    return state


print("Greeting node logging added successfully!")
```

```python
# Week 4 - Day 5
# Task 5 - Step 4: Add Logging to Intent Detection Node

def intent_detection_node(state):

    intent = state["intent"]

    state = log_transition(
        state,
        "intent_detection",
        f"Intent detected: {intent}"
    )

    return state


print("Intent detection logging added successfully!")
```

```python
# Week 4 - Day 5
# Task 5 - Step 5: Connect RAG Tool with RAG Node

def rag_node(state):

    query = state["conversation_history"][-1]

    rag_results = rag_search_tool(query)

    state["tool_outputs"]["rag_search"] = rag_results

    state = log_transition(
        state,
        "rag",
        "RAG search completed and results stored."
    )

    return state


print("RAG node connected successfully!")
```

```python
# Week 4 - Day 5
# Task 5 - Step 6: Connect Property Search with Recommendation Node

def recommendation_node(state):

    city = state["property_preferences"].get("city")
    area = state["property_preferences"].get("area")

    if city is None or area is None:

        state["tool_outputs"]["property_search"] = {
            "status": "clarification_needed",
            "message": "City and area are required."
        }

        state = log_transition(
            state,
            "recommendation",
            "City or area was missing, clarification is needed."
        )

        return state

    property_results = search_property_tool(
        city,
        area
    )

    state["tool_outputs"]["property_search"] = property_results

    state = log_transition(
        state,
        "recommendation",
        "Property search completed."
    )

    return state


print("Recommendation node connected successfully!")
```

```python
# Week 4 - Day 5
# Task 5 - Step 7: Connect Booking with Validation

def booking_node(state):

    city = state["property_preferences"].get("city")
    area = state["property_preferences"].get("area")

    appointment_date = state["tool_outputs"].get(
        "appointment_date"
    )

    appointment_time = state["tool_outputs"].get(
        "appointment_time"
    )

    property_search = state["tool_outputs"].get(
        "property_search",
        {}
    )

    property_results = property_search.get(
        "results",
        []
    )

    validation_result = validate_booking(
        city,
        area,
        appointment_date,
        appointment_time,
        property_results
    )

    state["tool_outputs"]["booking_validation"] = (
        validation_result
    )

    if validation_result["valid"]:

        state["appointment_status"] = "Booked"

        state = log_transition(
            state,
            "booking",
            "Booking validation passed and appointment was booked."
        )

    else:

        state["appointment_status"] = "Not Booked"

        state = log_transition(
            state,
            "booking",
            "Booking stopped because validation failed."
        )

    return state


print("Booking node connected with validation successfully!")
```

```python
# Week 4 - Day 5
# Task 5 - Step 8: Connect Email Tool with Email Node

def email_node(state):

    if state["appointment_status"] != "Booked":

        state = log_transition(
            state,
            "email",
            "Email was not prepared because appointment was not booked."
        )

        return state

    client_name = state["user_profile"].get(
        "name",
        "Customer"
    )

    property_name = (
        state["property_preferences"].get(
            "city",
            ""
        )
        + " "
        + state["property_preferences"].get(
            "area",
            ""
        )
        + " House"
    )

    appointment_date = state["tool_outputs"].get(
        "appointment_date"
    )

    appointment_time = state["tool_outputs"].get(
        "appointment_time"
    )

    employee = state["user_profile"].get(
        "employee",
        "Ahmed"
    )

    email_data = email_tool(
        client_name,
        property_name,
        appointment_date,
        appointment_time,
        employee
    )

    state["tool_outputs"]["email"] = email_data

    state = log_transition(
        state,
        "email",
        "Email data prepared for the assigned employee."
    )

    return state


print("Email node connected successfully!")
```

```python
# Week 4 - Day 6
# Setup - Restore Required Functions

import pandas as pd

# Load the existing real estate dataset
properties_df = pd.read_csv("zameen_properties_all.csv")


# Search properties from the real dataset
def search_properties(city, area):

    results = properties_df[
        (properties_df["city"].str.lower() == city.lower()) &
        (properties_df["area"].str.lower() == area.lower())
    ]

    return results


# Property search tool
def search_property_tool(city, area):

    results = search_properties(city, area)

    if len(results) == 0:

        return {
            "status": "not_found",
            "results": []
        }

    return {
        "status": "found",
        "results": results.to_dict(orient="records")
    }


# Check appointment availability
def availability_checker(appointment_date, appointment_time):

    if (
        appointment_date == "25 September 2026"
        and appointment_time == "4:00 PM"
    ):

        return {
            "available": True,
            "message": "Appointment slot is available."
        }

    else:

        return {
            "available": False,
            "message": "Appointment slot is not available."
        }


# Validate appointment
def validate_appointment(appointment_date, appointment_time):

    result = availability_checker(
        appointment_date,
        appointment_time
    )

    if result["available"]:

        return {
            "valid": True,
            "message": "Appointment slot is available."
        }

    return {
        "valid": False,
        "message": "Sorry, this appointment slot is not available. Please choose another time."
    }


print("Day 6 setup completed.")
print("Properties loaded:", len(properties_df))
```

```python
# Week 4 - Day 6
# Task 1 - Restore Day 5 Agent Nodes and Routing

from typing import TypedDict


# Agent State
class AgentState(TypedDict):

    conversation_history: list
    user_profile: dict
    property_preferences: dict
    budget: int
    intent: str
    tool_outputs: dict
    appointment_status: str
    execution_trace: list


# Log every node transition
def log_transition(state, node_name, message):

    state["execution_trace"].append({
        "node": node_name,
        "message": message
    })

    return state


# Intent routing
def route_intent(state):

    intent = state["intent"]

    if intent == "property_search":
        return "rag"

    elif intent == "appointment_booking":
        return "booking"

    elif intent == "rescheduling":
        return "rescheduling"

    elif intent == "cancellation":
        return "cancellation"

    else:
        return "goodbye"


# Greeting node
def greeting_node(state):

    state["conversation_history"].append(
        "Agent: Assalam-o-Alaikum, main aap ki property search mein help kar sakta hoon."
    )

    state = log_transition(
        state,
        "greeting",
        "Greeting message sent to customer."
    )

    return state


# Intent detection node
def intent_detection_node(state):

    intent = state["intent"]

    state = log_transition(
        state,
        "intent_detection",
        f"Intent detected: {intent}"
    )

    return state


# RAG node
def rag_node(state):

    query = state["conversation_history"][-1]

    rag_results = rag_search_tool(query)

    state["tool_outputs"]["rag_search"] = rag_results

    state = log_transition(
        state,
        "rag",
        "RAG search completed and results stored."
    )

    return state


# Recommendation node
def recommendation_node(state):

    city = state["property_preferences"].get("city")
    area = state["property_preferences"].get("area")

    if city is None or area is None:

        state["tool_outputs"]["property_search"] = {
            "status": "clarification_needed",
            "message": "City and area are required."
        }

        state = log_transition(
            state,
            "recommendation",
            "City or area was missing, clarification is needed."
        )

        return state

    property_results = search_property_tool(
        city,
        area
    )

    state["tool_outputs"]["property_search"] = property_results

    state = log_transition(
        state,
        "recommendation",
        "Property search completed."
    )

    return state


# Booking node
def booking_node(state):

    city = state["property_preferences"].get("city")
    area = state["property_preferences"].get("area")

    appointment_date = state["tool_outputs"].get(
        "appointment_date"
    )

    appointment_time = state["tool_outputs"].get(
        "appointment_time"
    )

    property_search = state["tool_outputs"].get(
        "property_search",
        {}
    )

    property_results = property_search.get(
        "results",
        []
    )

    validation_result = validate_booking(
        city,
        area,
        appointment_date,
        appointment_time,
        property_results
    )

    state["tool_outputs"]["booking_validation"] = validation_result

    if validation_result["valid"]:

        state["appointment_status"] = "Booked"

        state = log_transition(
            state,
            "booking",
            "Booking validation passed and appointment was booked."
        )

    else:

        state["appointment_status"] = "Not Booked"

        state = log_transition(
            state,
            "booking",
            "Booking stopped because validation failed."
        )

    return state


# Rescheduling node
def rescheduling_node(state):

    state["appointment_status"] = "Rescheduled"

    state = log_transition(
        state,
        "rescheduling",
        "Appointment rescheduling request processed."
    )

    return state


# Cancellation node
def cancellation_node(state):

    state["appointment_status"] = "Cancelled"

    state = log_transition(
        state,
        "cancellation",
        "Appointment cancellation request processed."
    )

    return state


# Email node
def email_node(state):

    if state["appointment_status"] != "Booked":

        state = log_transition(
            state,
            "email",
            "Email was not prepared because appointment was not booked."
        )

        return state

    client_name = state["user_profile"].get(
        "name",
        "Customer"
    )

    property_name = (
        state["property_preferences"].get("city", "")
        + " "
        + state["property_preferences"].get("area", "")
        + " House"
    )

    appointment_date = state["tool_outputs"].get(
        "appointment_date"
    )

    appointment_time = state["tool_outputs"].get(
        "appointment_time"
    )

    employee = state["user_profile"].get(
        "employee",
        "Ahmed"
    )

    email_data = email_tool(
        client_name,
        property_name,
        appointment_date,
        appointment_time,
        employee
    )

    state["tool_outputs"]["email"] = email_data

    state = log_transition(
        state,
        "email",
        "Email data prepared for the assigned employee."
    )

    return state


# Goodbye node
def goodbye_node(state):

    state["conversation_history"].append(
        "Agent: Thank you. Agar aap ko mazeed property assistance chahiye ho to batayein."
    )

    state = log_transition(
        state,
        "goodbye",
        "Conversation ended."
    )

    return state


print("All Day 5 agent nodes and routing restored successfully.")
```

```python
# Week 4 - Day 6
# Task 1 - Step 5G
# Restore Complete RAG Setup

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors

# Load property dataset
properties_df = pd.read_csv(
    "zameen_properties_all.csv"
)

# Create property documents
property_documents = []

for _, row in properties_df.iterrows():

    text = (
        f"Property ID: {row['property_id']}\n"
        f"City: {row['city']}\n"
        f"Property Type: {row['property_type']}\n"
        f"Title: {row['title']}\n"
        f"Price: {row['price']}\n"
        f"Bedrooms: {row['bedrooms']}\n"
        f"Bathrooms: {row['bathrooms']}\n"
        f"Area: {row['area']}\n"
        f"Description: {row['description']}\n"
        f"Agent: {row['agent_name']}\n"
        f"Amenities: {row['amenities']}\n"
        f"Availability: {row['availability']}"
    )

    property_documents.append(text)


# Create property chunks
property_chunks = []

for document in property_documents:

    words = document.split()

    for i in range(0, len(words), 100):

        chunk = " ".join(
            words[i:i + 100]
        )

        property_chunks.append(chunk)


# Load FAQ dataset
faqs_df = pd.read_csv(
    "zameen_faqs.csv"
)

# Create FAQ documents
faq_documents = []

for _, row in faqs_df.iterrows():

    text = (
        f"FAQ ID: {row['faq_id']}\n"
        f"Category: {row['category']}\n"
        f"Question: {row['question']}"
    )

    faq_documents.append(text)


# Create FAQ chunks
faq_chunks = []

for document in faq_documents:

    words = document.split()

    for i in range(0, len(words), 100):

        chunk = " ".join(
            words[i:i + 100]
        )

        faq_chunks.append(chunk)


# Combine all chunks
all_chunks = (
    property_chunks +
    faq_chunks
)


# Create TF-IDF vectorizer
vectorizer = TfidfVectorizer()

# Convert chunks into vectors
embeddings = vectorizer.fit_transform(
    all_chunks
)


# Create vector store
vector_store = NearestNeighbors(
    n_neighbors=5,
    metric="cosine"
)

# Store embeddings
vector_store.fit(
    embeddings
)


print("Complete RAG setup restored.")
print("Properties:", len(properties_df))
print("Property chunks:", len(property_chunks))
print("FAQ chunks:", len(faq_chunks))
print("Total chunks:", len(all_chunks))
print("Embedding shape:", embeddings.shape)
```
